```python
import numpy as np
import pandas as pd
import matplotlib
%matplotlib inline

import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("C:/MyProjects/BIA_assign/BIA_final/project_data.csv")
df.head()
df = df.drop('id', axis=1)
```

```python
# data type check
print(df.info())
```

```python
# null value check
print(df.isnull().sum())
```

```python
plt.figure(figsize=(8, 5))
sns.histplot(data=df,x='Age',bins=30,kde=True)

plt.title('Distribution of Customer Age')
plt.xlabel('Age')
plt.ylabel('Number of Customers')
plt.tight_layout()
plt.show()
```

```python
print(df['Annual_Income'].describe())
print(df['Annual_Income'].quantile([0.90, 0.95, 0.99, 0.999]))
print((df['Annual_Income'] > 300_000).sum())   # how many extreme outliers?
```

```python
plt.figure(figsize=(7, 5))

sns.countplot(
    data=df,
    x='Education',
    order=df['Education'].value_counts().index
)

plt.title('Distribution of Customers by Education')
plt.xlabel('Education')
plt.ylabel('Number of Customers')
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 5))

sns.countplot(
    data=df,
    x='Policy'
)

plt.title('Distribution of Insurance Policy Types')
plt.xlabel('Policy')
plt.ylabel('Number of Customers')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# income vs policy
# Create income bands
df['Income_Band'] = pd.cut(
    df['Annual_Income'],
    bins=[0, 100000, 200000, 300000, 500000, float('inf')],
    labels=[
        'Below 80K',
        '80K-1.5L',
        '1.5L-2L',
        '2L-3L',
        'Above 3L'
    ]
)

income_policy = pd.crosstab(
    df['Income_Band'],
    df['Policy'],
    normalize='index'
) * 100

print(income_policy.round(2))
```

```python
income_policy.plot(
    kind='bar',
    stacked=True,
    figsize=(12, 6)
)

plt.title('Policy Distribution Across Income Bands')
plt.xlabel('Income Band')
plt.ylabel('Percentage of Customers')
plt.legend(title='Policy', bbox_to_anchor=(1.05, 1))
plt.tight_layout()
plt.show()
```

```python
insurance_policy = pd.crosstab(
    df['Previously_Insured'],
    df['Policy'],
    normalize='index'
) * 100

print(insurance_policy.round(2))
insurance_policy.plot(
    kind='bar',
    stacked=True,
    figsize=(10, 6)
)

plt.title('Policy Distribution by Previous Insurance Status')
plt.xlabel('Previously Insured')
plt.ylabel('Percentage of Customers')
plt.legend(title='Policy', bbox_to_anchor=(1.05, 1))
plt.tight_layout()
plt.show()
```

```python
vehicle_age_policy = pd.crosstab(
    df['Vehicle_Age'],
    df['Policy'],
    normalize='index'
) * 100

print(vehicle_age_policy.round(2))
vehicle_age_policy.plot(
    kind='bar',
    stacked=True,
    figsize=(10, 6)
)

plt.title('Policy Distribution by Vehicle Age')
plt.xlabel('Vehicle Age')
plt.ylabel('Percentage of Customers')
plt.legend(title='Policy', bbox_to_anchor=(1.05, 1))
plt.tight_layout()
plt.show()
```

```python
damage_insurance_policy = pd.crosstab(
    [
        df['Vehicle_Damage'],
        df['Previously_Insured']
    ],
    df['Policy'],
    normalize='index'
) * 100

print(damage_insurance_policy.round(2))
damage_insurance_policy.plot(
    kind='bar',
    stacked=True,
    figsize=(12, 6)
)

plt.title('Policy Distribution by Vehicle Damage and Previous Insurance')
plt.xlabel('Vehicle Damage + Previously Insured')
plt.ylabel('Percentage of Customers')
plt.legend(title='Policy', bbox_to_anchor=(1.05, 1))
plt.tight_layout()
plt.show()
```
